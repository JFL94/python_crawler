```python
import time

def download_file(file_name):
    print(f"開始下載{file_name}...")
    time.sleep(2)   # 模擬下載花費時間
    print(f"{file_name}下載完成!")

def main_sync():
    start_time = time.time()
    download_file("檔案A.pdf")
    download_file("檔案B.pdf")
    download_file("檔案C.pdf")
    end_time = time.time()
    print(f"花費時間:{end_time -start_time:.2f}秒") 

if __name__ == "__main__":
    main_sync()
```

```python
import asyncio # python內建非同步函式庫
import nest_asyncio # 允許在已運行的事件迴圈中嵌套運行異步程式碼(巢狀event loop) .ipynb(內建event loop) .py檔不需要

nest_asyncio.apply() # 允許在ipynb運行異步程式碼

async def download_file(file_name):
    print(f"開始下載{file_name}...")
    await asyncio.sleep(2)   # 模擬下載花費時間
    print(f"{file_name}下載完成!")

async def main_async(): # asyncio.run()呼叫的非同步函式要用async def定義
    start_time = time.time()
    await asyncio.gather( # await 意思著等待非同步任務完成(一個任務包) asyncio.gather() 同時運行多個非同步任務
        download_file("檔案A.pdf"),
        download_file("檔案B.pdf"),
        download_file("檔案C.pdf")
    )
    end_time = time.time()
    print(f"花費時間:{end_time -start_time:.2f}秒") 

if __name__ == "__main__":
    asyncio.run(main_async()) # asyncio.run() 用於運行最高層級的非同步程式碼
```
